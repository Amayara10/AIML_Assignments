```python
import random

class VacuumCleaner:
    def __init__(self):

        # Different rooms in the house
        self.rooms = [
            "Living Room",
            "Bedroom",
            "Kitchen",
            "Study Room",
            "Dining Room"
        ]

        # Possible conditions of a room
        conditions = [
            "Clean",
            "Dirty",
            "Very Dirty",
            "Obstacle"
        ]

        # Randomly assign a condition to every room
        self.environment = {}

        for room in self.rooms:
            self.environment[room] = random.choice(conditions)

        # Vacuum starts from the first room
        self.location = 0

        # Battery starts at 100%
        self.battery = 100

        # Dustbin starts empty
        self.dustbin = 0

        # Maximum dustbin capacity
        self.dustbin_capacity = 100

        # Performance score
        self.score = 0

        # Count of completely cleaned rooms
        self.cleaned_rooms = 0

        # Count of partially cleaned rooms
        self.partial_rooms = 0

        # Store rooms where an obstacle blocked some portion
        self.obstacle_rooms = []


    # Display Initial Environment
    def show_environment(self):

        print("\nINITIAL HOUSE CONDITION")

        for room, condition in self.environment.items():
            print(room, ":", condition)


    # Clean Current Room
    def clean_room(self):

        # Find the current room
        room = self.rooms[self.location]

        # Find its condition
        condition = self.environment[room]

        print("\nCurrent Room:", room)
        print("Detected Condition:", condition)


        # CASE 1 - ROOM IS ALREADY CLEAN
        if condition == "Clean":

            print("Room is already clean.")
            print("Action: No cleaning required.")

            # Small score for correctly identifying clean room
            self.score += 2

            # Small battery consumption
            self.battery -= 2


        # CASE 2 - ROOM IS DIRTY
        elif condition == "Dirty":

            print("Room is Dirty.")
            print("Action: Normal cleaning started...")

            # Room becomes clean after cleaning
            self.environment[room] = "Clean"

            # Update score and counters
            self.score += 10
            self.cleaned_rooms += 1

            # Normal cleaning uses battery
            self.battery -= 10

            # Dust collected during cleaning
            self.dustbin += 30

            # Prevent dustbin from exceeding capacity
            if self.dustbin > self.dustbin_capacity:
                self.dustbin = self.dustbin_capacity

            print("Room cleaned successfully.")
            print("Dustbin Level:", self.dustbin, "%")


        # CASE 3 - ROOM IS VERY DIRTY
        elif condition == "Very Dirty":

            print("Room is VERY DIRTY.")
            print("Action: Deep cleaning started...")

            # Room becomes clean after deep cleaning
            self.environment[room] = "Clean"

            # Deep cleaning gives higher score
            self.score += 20
            self.cleaned_rooms += 1

            # Deep cleaning consumes more battery
            self.battery -= 20

            # More dust is collected during deep cleaning
            self.dustbin += 50

            # Prevent dustbin from exceeding capacity
            if self.dustbin > self.dustbin_capacity:
                self.dustbin = self.dustbin_capacity

            print("Deep cleaning completed successfully.")
            print("Dustbin Level:", self.dustbin, "%")


        # CASE 4 - OBSTACLE DETECTED
        elif condition == "Obstacle":

            print("\nOBSTACLE DETECTED!")
            print("Some portion of", room, "is blocked by an obstacle.")

            # Give warning to the user
            print("\nPlease remove the obstacle.")

            answer = input("Was the obstacle removed? (yes/no): ")


            # CASE 4A - OBSTACLE REMOVED
            if answer.lower() == "yes":

                print("\nObstacle removed successfully.")
                print("Vacuum can now access the entire room.")
                print("Cleaning the complete room...")

                # Entire room becomes clean
                self.environment[room] = "Clean"

                # Update score and counter
                self.score += 10
                self.cleaned_rooms += 1

                # Cleaning uses battery
                self.battery -= 10

                # Dust collected during cleaning
                self.dustbin += 30

                # Prevent dustbin from exceeding capacity
                if self.dustbin > self.dustbin_capacity:
                    self.dustbin = self.dustbin_capacity

                print("Complete room cleaned successfully.")
                print("Dustbin Level:", self.dustbin, "%")


            # CASE 4B - OBSTACLE NOT REMOVED
            else:

                print("\nObstacle was NOT removed.")
                print("Vacuum cannot enter the blocked portion.")

                # The vacuum does NOT skip the room.
                # It cleans the accessible/remaining portion.
                print("\nCleaning the remaining accessible portion...")

                print("Blocked portion will be left uncleaned.")

                # Store the room for final report
                self.obstacle_rooms.append(room)

                # Mark room as partially cleaned
                self.environment[room] = "Partially Cleaned"

                # Increase partial cleaning count
                self.partial_rooms += 1

                # Give score for cleaning accessible portion
                self.score += 7

                # Cleaning accessible area uses battery
                self.battery -= 10

                # Dust collected from accessible portion
                self.dustbin += 20

                # Prevent dustbin from exceeding capacity
                if self.dustbin > self.dustbin_capacity:
                    self.dustbin = self.dustbin_capacity

                print("Remaining portion cleaned successfully.")
                print("Blocked portion remains uncleaned.")
                print("Dustbin Level:", self.dustbin, "%")


    # Check Battery
    def check_battery(self):

        # Check if battery is low
        if self.battery <= 20:

            print("\nLOW BATTERY DETECTED!")
            print("Battery:", self.battery, "%")
            print("Vacuum is going to the charging station...")

            # Charge the vacuum
            self.charge_battery()

            return True

        return False


    # Charge Battery
    def charge_battery(self):

        print("\nCharging started...")

        # Restore battery to 100%
        self.battery = 100

        print("Battery charging completed.")
        print("Battery:", self.battery, "%")
        print("Vacuum is ready to continue cleaning.")


    # Check Dustbin
    def check_dustbin(self):

        # Check if dustbin is full
        if self.dustbin >= self.dustbin_capacity:

            print("\nDUSTBIN IS FULL!")
            print("Vacuum is going to the maintenance station...")

            # Empty the dustbin
            self.dustbin = 0

            print("Dustbin emptied successfully.")
            print("Dustbin Level:", self.dustbin, "%")
            print("Vacuum is ready to continue cleaning.")


    # Check All Rooms
    def check_all_rooms(self):

        # Check whether all rooms are clean
        all_clean = True

        for room, condition in self.environment.items():

            if condition != "Clean":
                all_clean = False
                break

        return all_clean


    # Wait For Two Hours
    def wait_two_hours(self):

        print("\nAll rooms are currently clean.")
        print("Vacuum is going into standby mode.")
        print("Vacuum will check the house again after 2 hours.")
        print("2 hours later...")
        print("Vacuum is ON again.")
        print("Checking all rooms again...")

        # Simulate changes while vacuum is sleeping
        for room in self.rooms:

            chance = random.choice([1, 2, 3, 4, 5])


            # Room becomes dirty
            if chance == 1:

                self.environment[room] = "Dirty"

                print(room, "has become Dirty again.")


            # Room becomes very dirty
            elif chance == 2:

                self.environment[room] = "Very Dirty"

                print(room, "has become Very Dirty again.")


            # Room remains clean
            else:

                print(room, "is still Clean.")


    # Move to Next Room
    def move(self):

        # Check if another room is available
        if self.location < len(self.rooms) - 1:

            # Move to next room
            self.location += 1

            # Moving consumes battery
            self.battery -= 5

            print("\nMoving to:", self.rooms[self.location])

        else:

            print("\nAll rooms have been visited.")


    # Run Vacuum Cleaner
    def run(self):

        print("SMART VACUUM CLEANER AGENT")


        # Show randomly generated environment
        self.show_environment()

        print("\nStarting Room:", self.rooms[self.location])
        print("Battery:", self.battery, "%")
        print("Dustbin:", self.dustbin, "%")


        # First cleaning cycle
        for i in range(len(self.rooms)):

            # Check battery before cleaning
            self.check_battery()

            # Check dustbin before cleaning
            self.check_dustbin()

            # Clean current room
            self.clean_room()

            print("Battery Remaining:", self.battery, "%")
            print("Dustbin Level:", self.dustbin, "%")


            # Move to next room
            if i < len(self.rooms) - 1:
                self.move()


        # Cleaning cycle is completed
        # Even if some room is partially cleaned,
        # vacuum will go into standby mode.
        print("\nCleaning cycle completed.")

        print("Vacuum is going into standby mode.")
        print("Vacuum will wake up after 2 hours.")


        # Simulate 2 hours
        print("\n2 hours later...")
        print("Vacuum is ON again.")
        print("Checking all rooms again...")


        # Reset location
        self.location = 0


        # Randomly check room conditions again
        for room in self.rooms:

            chance = random.choice([1, 2, 3, 4, 5])


            # Room becomes dirty
            if chance == 1:

                self.environment[room] = "Dirty"

                print(
                    room,
                    "has become Dirty again."
                )


            # Room becomes very dirty
            elif chance == 2:

                self.environment[room] = "Very Dirty"

                print(
                    room,
                    "has become Very Dirty again."
                )


            # Room remains in its current condition
            else:

                print(
                    room,
                    "condition remains:",
                    self.environment[room]
                )


        # Second cleaning cycle
        print("\nStarting second cleaning cycle...")


        for i in range(len(self.rooms)):

            # Check battery
            self.check_battery()

            # Check dustbin
            self.check_dustbin()

            # Clean current room
            self.clean_room()

            print(
                "Battery Remaining:",
                self.battery,
                "%"
            )

            print(
                "Dustbin Level:",
                self.dustbin,
                "%"
            )


            # Move to next room
            if i < len(self.rooms) - 1:
                self.move()


        # Final Report
        print("\nFINAL REPORT")

        print("\nFinal Environment:")

        for room, condition in self.environment.items():
            print(room, ":", condition)


        print(
            "\nCompletely Cleaned Rooms:",
            self.cleaned_rooms
        )

        print(
            "Partially Cleaned Rooms:",
            self.partial_rooms
        )

        print(
            "Performance Score:",
            self.score
        )

        print(
            "Battery Remaining:",
            self.battery,
            "%"
        )

        print(
            "Dustbin Level:",
            self.dustbin,
            "%"
        )


        # Obstacle Report
        if len(self.obstacle_rooms) > 0:

            print("\nOBSTACLE REPORT")

            for room in self.obstacle_rooms:

                print(
                    "WARNING:",
                    room,
                    ": Some portion was not cleaned because "
                    "an obstacle was not removed."
                )

        else:

            print(
                "\nNo room had an unresolved obstacle."
            )


# MAIN PROGRAM

# Create the Vacuum Cleaner Agent
vacuum = VacuumCleaner()

# Start the vacuum cleaner
vacuum.run()
```
